**24. Merging and concatenating (2 pts)**

Given:
```python
students = pd.DataFrame({"name": ["Ana", "Ben", "Carla", "Dan"],
                         "section": ["A", "B", "A", "B"]})
grades   = pd.DataFrame({"name": ["Ana", "Carla", "Dan", "Ella"],
                         "grade": [90, 85, 88, 92]})
```
- (a) Merge `students` and `grades` using an **inner** join. Print the result.
- (b) Merge `students` and `grades` using a **left** join. Print the result and add a comment explaining why Ben has `NaN`.
- (c) Use `pd.concat()` to add a new row for `"Ella"` in section `"A"` to `students`. Print the result.

````python
students = pd.DataFrame({"name": ["Ana", "Ben", "Carla", "Dan"],
                         "section": ["A", "B", "A", "B"]})
grades   = pd.DataFrame({"name": ["Ana", "Carla", "Dan", "Ella"],
                         "grade": [90, 85, 88, 92]})

# (a)
inner_merged = pd.merge(students, grades, on='name', how='inner')
print(inner_merged)

# (b)
left_merged = pd.merge(students, grades, on='name', how='left')
print(left_merged)
# Explanation: Ben has NaN for 'grade' because in a left join, all records from the left DataFrame (students)
# are kept, and since Ben does not exist in the right DataFrame (grades), the missing grade value is filled with NaN.

# (c)
new_row = pd.DataFrame({"name": ["Ella"], "section": ["A"]})
students = pd.concat([students, new_row], ignore_index=True)
print(students)
````

  ```python
  conn.execute('DROP TABLE IF EXISTS STUDENTS')
  ```

- (c) Create a table named `STUDENTS` with these columns, then print `"Table created successfully"`:

| Column | Type | Rule |
|---|---|---|
| `ID` | `INT` | `PRIMARY KEY`, `NOT NULL` |
| `NAME` | `TEXT` | `NOT NULL` |
| `AGE` | `INT` | `NOT NULL` |
| `COURSE` | `CHAR(50)` | |
| `GWA` | `REAL` | |

````python
import sqlite3

# (a)
conn = sqlite3.connect('school.db')
print("Opened database successfully")

# (b)
conn.execute('DROP TABLE IF EXISTS STUDENTS')

# (c)
conn.execute('''
CREATE TABLE STUDENTS (
    ID INT PRIMARY KEY NOT NULL,
    NAME TEXT NOT NULL,
    AGE INT NOT NULL,
    COURSE CHAR(50),
    GWA REAL
);
''')
print("Table created successfully")
````
